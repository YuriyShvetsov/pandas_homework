```python
import pandas as pd
rating = pd.read_csv('ratings.csv')
rating.head()
```

```python
rating_five = rating[ rating.rating == 5 ]
rating_five.head()
```

```python
rating_five = rating[['movieId', 'rating']][ rating.rating == 5 ]
rating_five.head()
```

```python
movies = pd.read_csv('movies.csv')
movies.head()
```

```python
rating_five.groupby('movieId').count().sort_values('rating', ascending=False).merge(movies, on='movieId', how='left').head(1)
```

```python
joined = rating_five.merge(movies, on='movieId', how='left')
joined.head()
```

```python
rating_five.groupby('movieId').count().sort_values('userId', ascending=False).merge(movies, on='movieId', how='left').head()
```

По данным файла power.csv посчитайте суммарное потребление стран Прибалтики (Латвия, Литва и Эстония) категорий 4, 12 и 21 за период с 2005 по 2010 год. Не учитывайте в расчётах отрицательные значения quantity.

```python
power = pd.read_csv('power.csv')
power.head()
```

```python
power_filtered = power[ 
    ((power['country'] == 'Latvia') | (power['country'] == 'Lithuania') | (power['country'] == 'Estonia')) & 
    ((power['category'] == 4) | (power['category'] == 12) | (power['category'] == 21)) &
    (power['quantity'] > 0) &
    ((power['year'] >= 2005) & (power['year'] <= 2010))
]
power_filtered.head()
```

```python
power_filtered['category'].unique()
```

```python
power_filtered['quantity'].sum()
```
